```python
#dbutils.fs.ls('/databricks-datasets/Rdatasets/data-001/csv/ggplot2/diamonds.csv')
csv_path = '/databricks-datasets/Rdatasets/data-001/csv/ggplot2/diamonds.csv'

from pyspark.sql.functions import col

df = (spark.read.format('csv')
      .option("header", True) 
      .option("InferSchema", True)
      .load(csv_path))

result_df = (df.select('cut', 'color', 'clarity', 'price', 'carat')
                .filter((col('cut') == 'Premium') & (col('price') > 5000))
)

display(result_df)

```

```python
%python 
employee_data = [  ('Alice',  'Engineering', 95000),
    ('Bob',    'Marketing',   72000),
    ('Carol',  'Engineering', 105000),
    ('David',  'HR',          65000),
    ('Eva',    'Engineering', 98000),]

columns = ['Name', 'Department', 'Salary']

emp_df = spark .createDataFrame(data=employee_data, schema=columns)
emp_df.createOrReplaceTempView("emp_table")

```

```python
%sql 

select department, count(*), avg(salary) as avg_salary, max(salary) as max_salary
from emp_table 
group by Department
order by avg_salary desc
```
